```sql
Create  TABLE IF NOT EXISTS gizmobox.gold.customer_address
AS
select 
  customers.customer_id,
  customers.customer_name,
  customers.date_of_birth,
  customers.email,
  customers.member_since,
  customers.telephone,
  addresses.shipping_address_line_1,
  addresses.shipping_city,
  addresses.shipping_state,
  addresses.shipping_postcode,
  addresses.billing_address_line_1,
  addresses.billing_city,
  addresses.billing_state,
  addresses.billing_postcode
from gizmobox.silver.customers as customers
INNER JOIN gizmobox.silver.addresses as addresses
ON customers.customer_id = addresses.customer_id
```
