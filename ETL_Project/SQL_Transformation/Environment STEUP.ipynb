Environment set-up for GizmoBox ETL project
1. dropping existing catalog
2. access cloud storage using %fs and abfss protocol
3. creating catalog - gizmobox
4. undergozmobox catalig creating schema landing, bronze, silver and gold.
5. under landing schema creating external volume on operational data

```sql
DROP CATALOG IF EXISTS gizmobox CASCADE;
DROP CATALOG IF EXISTS circuitbox  CASCADE;
```

```sql
%fs ls 'abfss://gizmobox@deacourseextlakehouse.dfs.core.windows.net/'
```

```sql
CREATE CATALOG IF NOT EXISTS gizmobox
MANAGED LOCATION 'abfss://gizmobox@deacourseextlakehouse.dfs.core.windows.net/'
COMMENT 'This is the GizmoBox catalog'
```

```sql
USE CATALOG gizmobox;

CREATE SCHEMA IF NOT EXISTS landing
MANAGED LOCATION 'abfss://gizmobox@deacourseextlakehouse.dfs.core.windows.net/landing'
COMMENT 'This is the landing zone schema';

CREATE SCHEMA IF NOT EXISTS bronze
MANAGED LOCATION 'abfss://gizmobox@deacourseextlakehouse.dfs.core.windows.net/bronze'
COMMENT 'This is the bronze zone schema';

CREATE SCHEMA IF NOT EXISTS silver
MANAGED LOCATION 'abfss://gizmobox@deacourseextlakehouse.dfs.core.windows.net/silver'
COMMENT 'This is the silver zone schema';

CREATE SCHEMA IF NOT EXISTS gold
MANAGED LOCATION 'abfss://gizmobox@deacourseextlakehouse.dfs.core.windows.net/gold'
COMMENT 'This is the gold zone schema';
```

```sql
USE SCHEMA landing;

CREATE EXTERNAL VOLUME IF NOT EXISTS operational_data
LOCATION 'abfss://gizmobox@deacourseextlakehouse.dfs.core.windows.net/landing/operational_data'
COMMENT 'This is the operational data volume';

```
