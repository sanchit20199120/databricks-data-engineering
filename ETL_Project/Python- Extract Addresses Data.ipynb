```python

addresses_df = (
  spark.read.
  format('csv').
  option('header','true').
  option('delimiter','\t').
  load('/Volumes/gizmobox/landing/operational_data/addresses/').
  select('*','_metadata.file_path','_metadata.file_modification_time')
)
display(addresses_df)
```

```python
addresses_df.writeTo('gizmobox.bronze.py_addresses').createOrReplace()
```
