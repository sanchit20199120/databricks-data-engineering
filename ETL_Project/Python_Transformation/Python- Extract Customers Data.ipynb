```python
customer_df = (
    spark.read.format('json').load('/Volumes/gizmobox/landing/operational_data/customers/').
              select('*','_metadata.file_path','_metadata.file_modification_time')
              )
display(customer_df)
```

```python
 
customer_df.write.format('delta').mode('overwrite').saveAsTable('gizmobox.bronze.py_customers')
```

```python
customer_df.writeTo('gizmobox.bronze.py_customers').createOrReplace();
```

```python
df = spark.table ('gizmobox.bronze.py_customers')
display(df)
```
