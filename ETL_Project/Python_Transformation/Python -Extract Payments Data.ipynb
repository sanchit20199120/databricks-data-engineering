```python
payments_df = spark.read.format('csv').option('delimiter',',').load('abfss://gizmobox@deacourseextlakehouse.dfs.core.windows.net/landing/external_data/payments')
display(payments_df)
```

```python

payment_schema = 'payment_id INTEGER,order_id INTEGER,payment_timestamp TIMESTAMP,payment_status INTEGER,payment_method STRING'
```

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, TimestampType

pyspark_payment_schema = StructType([
    StructField('payment_id', IntegerType()),
    StructField('order_id', IntegerType()),
    StructField('payment_timestamp', TimestampType()),
    StructField('payment_status', IntegerType()),
    StructField('payment_method', StringType())
])
```

```python
payments_df_transform = (
    spark.read.format('csv')
    .option('delimiter',',')
    .schema(pyspark_payment_schema)
    .load('abfss://gizmobox@deacourseextlakehouse.dfs.core.windows.net/landing/external_data/payments')
)
display(payments_df_transform)
```

```python
payments_df_transform.writeTo('gizmobox.bronze.py_payments').createOrReplace()
```
