```sql
Create Table if not exists gizmobox.gold.order_summary_monthly
AS
select customer_id,
date_format(order_date,'yyyy-MM') AS month,
count(distinct order_id) as total_orders,
 sum(item_quantity) as total_item_bought,
  sum(item_price * item_quantity) as Total_amount_spent 
from gizmobox.silver.orders
group by customer_id, date_format(order_date,'yyyy-MM');
```
