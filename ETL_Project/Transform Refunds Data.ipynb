```sql
CREATE OR REPLACE TABLE gizmobox.silver.refunds
AS
select 
refund_id,
payment_id,
date_format(refund_timestamp,'yyyy-MM-dd')::DATE as refund_date,
date_format(refund_timestamp,'HH:mm:ss') as refund_timestamp,
refund_amount,
split(refund_reason, ':')[0] as refund_reason,
split(refund_reason, ':')[1] as refund_source
from sql_gizmobox_db_catalog_sqlserver.dbo.refunds;
```
