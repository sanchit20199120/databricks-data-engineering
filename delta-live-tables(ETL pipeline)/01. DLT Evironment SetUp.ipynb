### Environment Setup: Create below Metastore objects under unity Catalog
1. External Location
2. Catalog
3. schema
4. volume


```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS dea_course_ext_dl_circuitbox
URL 'abfss://circuitbox@deacourseextlakehouse.dfs.core.windows.net/'
WITH
(STORAGE CREDENTIAL dea_course_ext_sc)
COMMENT 'EXTERNAL LOCATION ACCESS TO DATABRICKS USING STORAGE CREDENTIAL FOR PROJECT CIRCUITBOX';
```

```sql
%fs ls 'abfss://circuitbox@deacourseextlakehouse.dfs.core.windows.net/'
```

```sql
CREATE CATALOG IF NOT EXISTS CIRCUITBOX
MANAGED LOCATION 'abfss://circuitbox@deacourseextlakehouse.dfs.core.windows.net/';
```

```sql
SHOW CATALOGS;
```

```sql
USE CATALOG CIRCUITBOX;

CREATE SCHEMA IF NOT EXISTS landing
MANAGED LOCATION 'abfss://circuitbox@deacourseextlakehouse.dfs.core.windows.net/landing'
COMMENT 'FOR STAGING FILES IN LANDING ZONE';

CREATE SCHEMA IF NOT EXISTS bronze
MANAGED LOCATION 'abfss://circuitbox@deacourseextlakehouse.dfs.core.windows.net/bronze'
COMMENT 'FOR ETL FILES IN LAKEHOUSE';

CREATE SCHEMA IF NOT EXISTS silver
MANAGED LOCATION 'abfss://circuitbox@deacourseextlakehouse.dfs.core.windows.net/silver'
COMMENT 'FOR ETL FILES IN LAKEHOUSE';

CREATE SCHEMA IF NOT EXISTS gold
MANAGED LOCATION 'abfss://circuitbox@deacourseextlakehouse.dfs.core.windows.net/gold'
COMMENT 'FOR ETL FILES IN LAKEHOUSE';

CREATE SCHEMA IF NOT EXISTS lakehouse
MANAGED LOCATION 'abfss://circuitbox@deacourseextlakehouse.dfs.core.windows.net/lakehouse'
COMMENT 'FOR ETL FILES IN LAKEHOUSE';
```

```sql
SHOW SCHEMAS;
```

```sql
USE CATALOG CIRCUITBOX;
USE SCHEMA landing;
```

```sql
CREATE EXTERNAL VOLUME IF NOT EXISTS operational_data
LOCATION 'abfss://circuitbox@deacourseextlakehouse.dfs.core.windows.net/landing/operational_data'
COMMENT 'EXTERNAL VOLUME FOR STAGING FILES IN LANDING ZONE';
```
