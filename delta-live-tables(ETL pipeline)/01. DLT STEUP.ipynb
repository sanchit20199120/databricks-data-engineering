### Environment Setup
1. Catalog
2. schema
3. volume


```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS dea_course_ext_dl_circuitbox
URL 'abfss://circuitbox@deacourseextlakehouse.dfs.core.windows.net/'
WITH
(STORAGE CREDENTIAL dea_course_ext_sc)
COMMENT 'EXTERNAL LOCATION ACCESS TO DATABRICKS USING STORAGE CREDENTIAL FOR PROJECT CIRCUITBOX';
```

```sql
%fs ls 'abfss://circuitbox@deacourseextlakehouse.dfs.core.windows.net/'
```

```sql
CREATE CATALOG IF NOT EXISTS CIRCUITBOX
MANAGED LOCATION 'abfss://circuitbox@deacourseextlakehouse.dfs.core.windows.net/circuitbox';
```

```sql
SHOW CATALOGS;
```

```sql
USE CATALOG CIRCUITBOX;

CREATE SCHEMA IF NOT EXISTS landing
MANAGED LOCATION 'abfss://circuitbox@deacourseextlakehouse.dfs.core.windows.net/circuitbox/landing'
COMMENT 'FOR STAGING FILES IN LANDING ZONE';



CREATE SCHEMA IF NOT EXISTS lakehouse
MANAGED LOCATION 'abfss://circuitbox@deacourseextlakehouse.dfs.core.windows.net/circuitbox/lakehouse'
COMMENT 'FOR ETL FILES IN LAKEHOUSE';
```

```sql
USE CATALOG CIRCUITBOX;
```

```sql
SHOW SCHEMAS;
```

```sql
USE CATALOG CIRCUITBOX;
USE SCHEMA landing;
```

```sql
CREATE EXTERNAL VOLUME IF NOT EXISTS operational_data
LOCATION 'abfss://circuitbox@deacourseextlakehouse.dfs.core.windows.net/landing/operational_data'
COMMENT 'EXTERNAL VOLUME FOR STAGING FILES IN LANDING ZONE';
```
